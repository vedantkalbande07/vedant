```python
import seaborn as sns
import numpy as np
import pandas as pd

# a) Load the Titanic dataset
titanic = sns.load_dataset('titanic')

# b) Extract Age as a NumPy array (removing missing values)
age_array = titanic['age'].dropna().to_numpy()

print("--- Part B ---")
print(f"First 10 values: {age_array[:10]}")
print(f"Shape of the array: {age_array.shape}")

# c) Create a multi-dimensional NumPy array
# We drop rows where any of these specific columns have NaN
subset = titanic[['age', 'fare', 'sibsp', 'parch']].dropna()
multi_array = subset.to_numpy()

print("\n--- Part C ---")
print(f"Shape of multi-dimensional array: {multi_array.shape}")

# d) Create a new DataFrame from the array
new_df = pd.DataFrame(multi_array, columns=['age', 'fare', 'sibsp', 'parch'])

print("\n--- Part D ---")
print("First 5 rows of the new DataFrame:")
print(new_df.head())
```
